```python
# ============================================================
# AI E-COMMERCE SALES ANALYTICS ASSISTANT
# Google Colab + Groq + Pandas
# ============================================================

# STEP 1: Install libraries
!pip install -q groq pandas matplotlib

# STEP 2: Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from groq import Groq
from google.colab import userdata

# ============================================================
# STEP 3: Connect to Groq
# ============================================================

api_key = userdata.get("GROQ_API_KEY")

client = Groq(api_key=api_key)

print("✅ Groq API connected successfully!")


# ============================================================
# STEP 4: Create E-Commerce Dataset
# ============================================================

data = {
    "Product": [
        "Laptop", "Mobile", "Headphones", "Smartwatch",
        "Laptop", "Mobile", "Headphones", "Smartwatch",
        "Laptop", "Mobile", "Headphones", "Smartwatch"
    ],

    "Category": [
        "Electronics", "Electronics", "Accessories", "Accessories",
        "Electronics", "Electronics", "Accessories", "Accessories",
        "Electronics", "Electronics", "Accessories", "Accessories"
    ],

    "Region": [
        "North", "South", "East", "West",
        "North", "South", "East", "West",
        "South", "North", "West", "East"
    ],

    "Sales": [
        85000, 60000, 15000, 22000,
        90000, 65000, 18000, 25000,
        78000, 55000, 17000, 24000
    ],

    "Quantity": [
        10, 15, 30, 20,
        11, 16, 35, 22,
        9, 14, 32, 21
    ],

    "Profit": [
        12000, 9000, 4500, 6000,
        14000, 10000, 5000, 6500,
        11000, 8000, 4800, 6200
    ]
}

df = pd.DataFrame(data)

print("\n📊 E-Commerce Dataset")
display(df)


# ============================================================
# STEP 5: Basic Analytics
# ============================================================

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_quantity = df["Quantity"].sum()

print("\n========== BUSINESS SUMMARY ==========")

print("💰 Total Sales:", total_sales)
print("💵 Total Profit:", total_profit)
print("📦 Total Quantity Sold:", total_quantity)


# ============================================================
# STEP 6: Product Analysis
# ============================================================

product_sales = (
    df.groupby("Product")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\n========== SALES BY PRODUCT ==========")
print(product_sales)


# ============================================================
# STEP 7: Region Analysis
# ============================================================

region_sales = (
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\n========== SALES BY REGION ==========")
print(region_sales)


# ============================================================
# STEP 8: Profit by Product
# ============================================================

product_profit = (
    df.groupby("Product")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

print("\n========== PROFIT BY PRODUCT ==========")
print(product_profit)


# ============================================================
# STEP 9: Sales Chart
# ============================================================

plt.figure(figsize=(8, 5))

product_sales.plot(kind="bar")

plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ============================================================
# STEP 10: Region Chart
# ============================================================

plt.figure(figsize=(8, 5))

region_sales.plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.tight_layout()
plt.show()


# ============================================================
# STEP 11: Prepare Data for AI
# ============================================================

summary = f"""
E-COMMERCE SALES DATA

Total Sales: ₹{total_sales:,}

Total Profit: ₹{total_profit:,}

Total Quantity Sold: {total_quantity}

Sales by Product:
{product_sales.to_string()}

Profit by Product:
{product_profit.to_string()}

Sales by Region:
{region_sales.to_string()}
"""


# ============================================================
# STEP 12: AI Business Analysis
# ============================================================

prompt = f"""
You are an AI E-Commerce Sales Analytics Assistant.

Analyze the following business data:

{summary}

Provide the analysis in simple language.

Include:

1. Overall business performance
2. Best-selling product
3. Most profitable product
4. Best-performing region
5. Important sales trends
6. Areas that need improvement
7. Three practical business recommendations

Do not make up information that is not present in the data.
"""

response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {
            "role": "system",
            "content": "You are an expert e-commerce data analyst."
        },
        {
            "role": "user",
            "content": prompt
        }
    ],
    temperature=0.2
)

print("\n\n🤖 AI BUSINESS ANALYSIS")
print("=" * 50)

print(response.choices[0].message.content)


# ============================================================
# STEP 13: Ask AI Questions About Your Data
# ============================================================

print("\n\n======================================")
print("🤖 AI E-COMMERCE DATA ASSISTANT")
print("======================================")

question = input("\nAsk a question about your sales data: ")

data_for_ai = df.to_string(index=False)

question_prompt = f"""
You are an AI E-Commerce Data Analytics Assistant.

Here is the complete sales dataset:

{data_for_ai}

Answer the user's question using ONLY the information
available in the dataset.

User Question:
{question}

Give a simple and accurate answer.

If calculations are required, calculate them using the data.
"""

response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {
            "role": "system",
            "content": "You are an expert data analyst."
        },
        {
            "role": "user",
            "content": question_prompt
        }
    ],
    temperature=0.1
)

print("\n🤖 AI Answer:")
print(response.choices[0].message.content)
```
